```python
    import h5py

from IPython.core.display_functions import display
from auroraPSI.h5_source_adapter import H5Adapter
```

```python
import numpy as np
filename = "../data/longer_data.h5"
```

```python
h5_adapter = H5Adapter.from_h5(filename)
h5_adapter.X
```

```python

with h5py.File(filename, "r") as result_file:
    display(result_file)
    display(result_file.keys())
    display(result_file['G'].keys())
    display(result_file['G']['axis0'])
    display(result_file['G']['axis1'])
    display(np.array(result_file['G']['block0_items']))
    display(np.array(result_file['G']['block0_values']))
    display(np.array(result_file['G']['axis0']))
    display(np.array(result_file['G']['axis1']))
    column_names = np.array(result_file['G']['axis0'])
    data = np.array(result_file['G']['block0_values'])
    times = np.array(result_file['G']['axis1'])
```

```python
import pandas as pd

test_G = pd.DataFrame(columns = column_names, index = times, data=data)
test_G.iloc[:,2].plot()
```
